```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/villes_france_lat_long.csv')

# Créer un GeoDataFrame
gdf = gpd.GeoDataFrame(
    df, 
    geometry=gpd.points_from_xy(df.Longitude, df.Latitude),
    crs='EPSG:4326'
)

plt.show()
# Visualiser
fig, ax = plt.subplots(figsize=(10, 10))
gdf.plot(ax=ax, color='red', markersize=50)

# Ajouter les noms des villes
for idx, row in gdf.iterrows():
    ax.annotate(row['Ville'], xy=(row.Longitude, row.Latitude), 
                xytext=(3, 3), textcoords='offset points', fontsize=8)

plt.title('Villes de France')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.tight_layout()
plt.show()
```

```python
MARSEILLE = (43.2965, 5.3698)
def get_cities(path):
    df = pd.read_csv(path)
    position_cities = list(zip(df["Latitude"], df["Longitude"]))
    position_cities.remove(MARSEILLE) # Retire Marseille des villes à visiter
    return position_cities


PATH = '../data/villes_france_lat_long.csv'
CITIES = get_cities(PATH)
CITIES
```
